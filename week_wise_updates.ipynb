### What does success look like
**Week 1 — Define the Contract of “Any Circuit”**

**Goal:** Prevent ambiguity later.

  * QASM-like
  * Python circuit object
* Decide **logical gate set**:

  * `{H, X, Z, S, T, CNOT, Measure}`

* One page design note:

  * “What circuits my compiler accepts”
* Parser → MLIR Logical IR

* You can feed *any small circuit* and get valid MLIR
