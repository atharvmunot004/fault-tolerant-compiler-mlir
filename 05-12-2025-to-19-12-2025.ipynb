## Coherent vs Incoherent Errors in Quantum Systems


### Overview

Understanding the difference between **coherent** and **incoherent** errors is crucial for quantum error correction. These two types of errors have fundamentally different characteristics and require different approaches to mitigate.

---

### Coherent Errors

**Coherent errors** are **systematic, deterministic errors** that preserve quantum coherence. They arise from imperfect control of quantum gates and can be described as **unitary rotations** away from the intended operation.

#### Key Characteristics:

1. **Deterministic**: The error is the same every time (not random)
2. **Unitary**: Can be described by a unitary operator (e.g., rotation)
3. **Preserves coherence**: Maintains quantum superposition and entanglement
4. **Cumulative**: Errors can build up coherently over time
5. **Reversible**: In principle, can be corrected by applying the inverse operation

#### Examples:

- **Over-rotation in gates**: A CNOT gate that rotates slightly more than intended
  - Intended: `CNOT`
  - Actual: `CNOT · R_z(θ)` where `θ` is a small angle
- **Calibration errors**: Systematic miscalibration of gate parameters
- **Control field errors**: Imperfect control pulses in quantum gates
- **Crosstalk**: Unwanted interactions between qubits during gate operations

#### Mathematical Description:

A coherent error can be represented as:
```
U_actual = U_intended · U_error
```
where `U_error` is a unitary operator close to the identity.

For example, a coherent bit-flip error might be:
```
U_error = exp(-iθX/2) ≈ I - iθX/2
```
where `θ` is small and `X` is the Pauli-X operator.

---

### Incoherent Errors

**Incoherent errors** are **random, stochastic errors** that destroy quantum coherence. They arise from interactions with the environment and can be described using **quantum channels** (completely positive trace-preserving maps).

#### Key Characteristics:

1. **Random/Stochastic**: The error varies randomly each time
2. **Non-unitary**: Described by quantum channels (Kraus operators)
3. **Destroys coherence**: Causes decoherence and loss of quantum information
4. **Probabilistic**: Errors occur with certain probabilities
5. **Irreversible**: Cannot be perfectly reversed (information is lost to environment)

#### Examples:

- **Bit-flip errors**: Random X errors (|0⟩ ↔ |1⟩)
- **Phase-flip errors**: Random Z errors (|+⟩ ↔ |-⟩)
- **Depolarizing noise**: Random application of X, Y, or Z with equal probability
- **Amplitude damping**: Energy loss to the environment (|1⟩ → |0⟩)
- **Phase damping**: Loss of phase information without energy loss
- **Measurement errors**: Imperfect readout of qubit states

#### Mathematical Description:

Incoherent errors are described by **quantum channels** using Kraus operators:

```
ρ → Σᵢ Kᵢ ρ Kᵢ†
```

where `Σᵢ Kᵢ† Kᵢ = I` (trace-preserving condition).

For example, a bit-flip error with probability `p`:
```
K₀ = √(1-p) I
K₁ = √p X
```

---

### Comparison Table

| Property | Coherent Errors | Incoherent Errors |
|----------|----------------|-------------------|
| **Nature** | Systematic, deterministic | Random, stochastic |
| **Description** | Unitary operators | Quantum channels (Kraus operators) |
| **Coherence** | Preserves coherence | Destroys coherence |
| **Predictability** | Predictable, same every time | Unpredictable, varies each time |
| **Cumulative** | Can build up coherently | Independent occurrences |
| **Reversibility** | In principle reversible | Irreversible (information loss) |
| **Source** | Control imperfections | Environmental interactions |
| **Example** | Over-rotation in gates | Bit-flip, phase-flip noise |

---

### Why This Distinction Matters

#### For Quantum Error Correction:

1. **Coherent Errors**:
   - Can be corrected using **compilation techniques** (e.g., gate decomposition)
   - May require **calibration** and **characterization**
   - Can be mitigated through **pulse shaping** and **optimal control**
   - Some QEC codes can handle small coherent errors

2. **Incoherent Errors**:
   - Require **quantum error correction codes** (like the 3-qubit repetition code)
   - Need **syndrome measurement** and **error correction**
   - Can be described by **error models** (e.g., depolarizing channel)
   - Are the primary target of most QEC protocols

#### Real-World Quantum Systems:

- **NISQ devices** (Noisy Intermediate-Scale Quantum): Experience both types
  - Coherent: Gate calibration errors, crosstalk
  - Incoherent: Decoherence, thermal noise, measurement errors
- **Fault-tolerant quantum computing**: Must handle both types
  - Coherent errors: Addressed through gate compilation and calibration
  - Incoherent errors: Addressed through QEC codes and error correction

---

### Error Models in Practice

#### Common Incoherent Error Models:

1. **Depolarizing Channel**:
   ```
   With probability p: apply random Pauli (X, Y, or Z)
   With probability 1-p: no error
   ```

2. **Bit-Flip Channel**:
   ```
   With probability p: apply X (bit-flip)
   With probability 1-p: no error
   ```

3. **Phase-Flip Channel**:
   ```
   With probability p: apply Z (phase-flip)
   With probability 1-p: no error
   ```

#### Coherent Error Models:

1. **Rotation Error**:
   ```
   Apply R_z(θ) instead of I, where θ is a small angle
   ```

2. **Over-rotation**:
   ```
   Apply CNOT with extra rotation: CNOT · R_x(ε)
   ```

---

### Mitigation Strategies

#### For Coherent Errors:

1. **Gate Calibration**: Regularly calibrate gate parameters
2. **Optimal Control**: Use shaped pulses to minimize errors
3. **Compilation**: Decompose gates to minimize error accumulation
4. **Characterization**: Use quantum process tomography to identify errors

#### For Incoherent Errors:

1. **Quantum Error Correction**: Use QEC codes (repetition codes, surface codes, etc.)
2. **Error Detection**: Syndrome measurement to identify errors
3. **Error Correction**: Apply correction operations based on syndrome
4. **Fault Tolerance**: Design circuits that are robust to errors

---

### Key Takeaways

1. **Coherent errors** are systematic and preserve quantum coherence; they can be mitigated through calibration and control.
2. **Incoherent errors** are random and destroy coherence; they require quantum error correction codes.
3. Real quantum systems experience **both types** of errors simultaneously.
4. Effective quantum error correction must address **both coherent and incoherent errors**.
5. The distinction is fundamental to understanding how quantum information degrades in noisy systems.

## Quantum Process Tomography (QPT)

### Overview

**Quantum Process Tomography (QPT)** is a technique used to **characterize and reconstruct** an unknown quantum operation (gate or channel) by measuring its effect on a complete set of input states. It's the quantum analog of classical process characterization and is essential for:

- **Gate calibration**: Understanding how gates actually behave vs. how they're supposed to behave
- **Error characterization**: Identifying and quantifying errors in quantum operations
- **Device benchmarking**: Assessing the quality of quantum hardware
- **Error mitigation**: Providing information needed to correct coherent errors

---

### What is Process Tomography?

Just as **quantum state tomography** reconstructs an unknown quantum state, **quantum process tomography** reconstructs an unknown quantum operation (process/channel).

#### The Problem:

We have a quantum gate/operation `ε` that we want to characterize:
- **Ideal operation**: `U` (e.g., a CNOT gate)
- **Actual operation**: `ε` (unknown, may have errors)
- **Goal**: Determine what `ε` actually does

#### The Solution:

1. Prepare a **complete set of input states** (tomographically complete basis)
2. Apply the unknown process `ε` to each input state
3. Perform **state tomography** on the output states
4. Reconstruct the process from the measurement data

---

### Mathematical Foundation

#### Process Representation:

A quantum process `ε` can be represented in several ways:

1. **Kraus Representation** (for general quantum channels):
   ```
   ε(ρ) = Σᵢ Kᵢ ρ Kᵢ†
   ```
   where `Σᵢ Kᵢ† Kᵢ = I` (trace-preserving)

2. **Choi-Jamiolkowski Representation**:
   ```
   C_ε = (I ⊗ ε)(|Φ⁺⟩⟨Φ⁺|)
   ```
   where `|Φ⁺⟩` is a maximally entangled state

3. **Process Matrix (χ-matrix)**:
   ```
   ε(ρ) = Σᵢⱼ χᵢⱼ Eᵢ ρ Eⱼ†
   ```
   where `{Eᵢ}` is a basis for operators (e.g., Pauli basis)

#### For a Single Qubit:

A single-qubit process can be described by a **4×4 process matrix** `χ` in the Pauli basis:
- Basis: `{I, X, Y, Z}`
- The process matrix has 16 real parameters (with constraints)

---

### How Quantum Process Tomography Works

#### Step-by-Step Process:

1. **Choose Input States**:
   - For 1 qubit: Need at least 4 linearly independent states
   - Common choice: `{|0⟩, |1⟩, |+⟩, |+i⟩}` or `{|0⟩, |1⟩, |+⟩, |-⟩}`
   - For n qubits: Need 4ⁿ input states

2. **Prepare Each Input State**:
   - Create the desired input state using gates

3. **Apply Unknown Process**:
   - Apply the gate/operation we want to characterize

4. **Measure Output States**:
   - Perform state tomography on each output
   - Measure in different bases (X, Y, Z for qubits)

5. **Reconstruct Process Matrix**:
   - Use linear algebra to reconstruct the process matrix `χ`
   - This matrix completely describes the quantum operation

6. **Analyze Results**:
   - Compare with ideal process
   - Extract error parameters (fidelity, process fidelity, etc.)

---

### Types of Process Tomography

#### 1. **Standard Process Tomography**:
   - Full reconstruction of the process matrix
   - Requires many measurements (exponential in qubit number)
   - Provides complete information about the process

#### 2. **Gate Set Tomography (GST)**:
   - Characterizes entire gate sets simultaneously
   - More efficient than standard QPT
   - Accounts for state preparation and measurement (SPAM) errors

#### 3. **Randomized Benchmarking**:
   - Estimates average gate fidelity
   - More scalable than full tomography
   - Doesn't provide full process information

---

### Key Metrics from Process Tomography

#### 1. **Process Fidelity**:
   ```
   F_proc = Tr(χ_ideal · χ_actual)
   ```
   - Measures how close the actual process is to the ideal
   - Range: 0 to 1 (1 = perfect)

#### 2. **Average Gate Fidelity**:
   ```
   F_avg = ∫ dψ ⟨ψ|U† ε(|ψ⟩⟨ψ|) U|ψ⟩
   ```
   - Average fidelity over all input states
   - Related to process fidelity

#### 3. **Diamond Norm**:
   ```
   ||ε - U||_⋄ = max_ρ ||(ε ⊗ I)(ρ) - (U ⊗ I)(ρ)||_1
   ```
   - Worst-case error measure
   - Useful for fault-tolerant threshold calculations

---

### Applications in Quantum Error Correction

#### 1. **Characterizing Coherent Errors**:
   - Identify systematic over-rotations
   - Measure calibration errors
   - Detect crosstalk between qubits

#### 2. **Gate Calibration**:
   - Adjust gate parameters based on tomography results
   - Improve gate fidelity iteratively

#### 3. **Error Model Construction**:
   - Build accurate error models for simulations
   - Design better error correction codes

#### 4. **Device Characterization**:
   - Benchmark quantum hardware
   - Compare different quantum processors

---

### Challenges and Limitations

#### 1. **Exponential Scaling**:
   - For n qubits, need ~4ⁿ measurements
   - Becomes impractical for large systems

#### 2. **SPAM Errors**:
   - State preparation and measurement errors affect results
   - Need to account for these in analysis

#### 3. **Statistical Errors**:
   - Finite number of measurements → statistical uncertainty
   - Need many shots for accurate reconstruction

#### 4. **Computational Cost**:
   - Reconstructing process matrix requires significant computation
   - Matrix inversion and optimization needed

---

### Example: Characterizing a Single-Qubit Gate

Let's see how process tomography works in practice for a simple gate.


````python
# Quantum Process Tomography Example: Characterizing a Single-Qubit Gate

from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector, Operator, process_fidelity
from qiskit_aer import AerSimulator
import numpy as np

print("="*70)
print("QUANTUM PROCESS TOMOGRAPHY: Characterizing a Single-Qubit Gate")
print("="*70)

# Define the ideal gate: Hadamard gate
ideal_gate = Operator([[1/np.sqrt(2), 1/np.sqrt(2)],
                       [1/np.sqrt(2), -1/np.sqrt(2)]])

print("\n1. IDEAL HADAMARD GATE:")
print(f"   Matrix representation:\n{ideal_gate.data}")

# Create a circuit with the ideal gate
circ_ideal = QuantumCircuit(1)
circ_ideal.unitary(ideal_gate, [0])

print("\n2. SIMULATING PROCESS TOMOGRAPHY:")
print("   (In practice, this would involve many measurements)")

# For demonstration, we'll show what input states are needed
print("\n   Required input states for 1-qubit QPT:")
print("   - |0⟩ (computational basis)")
print("   - |1⟩ (computational basis)")
print("   - |+⟩ = (|0⟩ + |1⟩)/√2 (X basis)")
print("   - |+i⟩ = (|0⟩ + i|1⟩)/√2 (Y basis)")

# Create these input states
input_states = {
    '|0⟩': QuantumCircuit(1),  # Already in |0⟩
    '|1⟩': QuantumCircuit(1),
    '|+⟩': QuantumCircuit(1),
    '|+i⟩': QuantumCircuit(1)
}
input_states['|1⟩'].x(0)
input_states['|+⟩'].h(0)
input_states['|+i⟩'].h(0)
input_states['|+i⟩'].s(0)  # S gate creates |+i⟩

print("\n3. APPLYING GATE TO EACH INPUT STATE:")

for name, circ in input_states.items():
    # Apply the gate
    circ_gate = circ.copy()
    circ_gate.unitary(ideal_gate, [0])
    
    # Get output state
    output_state = Statevector.from_instruction(circ_gate)
    print(f"   {name} → {output_state}")

print("\n4. PROCESS FIDELITY:")
print("   (In real QPT, we'd reconstruct the process matrix and compute fidelity)")
print("   Process fidelity measures how close the actual process is to ideal")
print("   F_proc = Tr(χ_ideal · χ_actual)")

print("\n" + "="*70)
print("NOTE: Full QPT requires many measurements and statistical analysis")
print("This is a simplified demonstration of the concept")
print("="*70)
````

### Example: Characterizing a Gate with Errors

Let's see how process tomography can identify errors in a gate.


````python
# Characterizing a Gate with Coherent Errors

from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector, Operator
import numpy as np

print("="*70)
print("PROCESS TOMOGRAPHY: Detecting Coherent Errors")
print("="*70)

# Ideal rotation: R_y(π/2) = Y^(1/2)
ideal_angle = np.pi / 2
ideal_gate = Operator([
    [np.cos(ideal_angle/2), -np.sin(ideal_angle/2)],
    [np.sin(ideal_angle/2), np.cos(ideal_angle/2)]
])

# Actual gate with coherent error: over-rotation by 5%
error_angle = ideal_angle * 1.05  # 5% over-rotation
actual_gate = Operator([
    [np.cos(error_angle/2), -np.sin(error_angle/2)],
    [np.sin(error_angle/2), np.cos(error_angle/2)]
])

print("\n1. IDEAL GATE: R_y(π/2)")
print(f"   Rotation angle: {ideal_angle:.4f} radians ({np.degrees(ideal_angle):.2f}°)")
print(f"   Matrix:\n{ideal_gate.data}")

print("\n2. ACTUAL GATE: R_y(π/2) with 5% over-rotation")
print(f"   Rotation angle: {error_angle:.4f} radians ({np.degrees(error_angle):.2f}°)")
print(f"   Error: {np.degrees(error_angle - ideal_angle):.2f}° over-rotation")
print(f"   Matrix:\n{actual_gate.data}")

print("\n3. TESTING ON DIFFERENT INPUT STATES:")

test_states = {
    '|0⟩': QuantumCircuit(1),
    '|1⟩': QuantumCircuit(1),
    '|+⟩': QuantumCircuit(1),
    '|+i⟩': QuantumCircuit(1)
}
test_states['|1⟩'].x(0)
test_states['|+⟩'].h(0)
test_states['|+i⟩'].h(0)
test_states['|+i⟩'].s(0)

print("\n   Input State | Ideal Output | Actual Output | Difference")
print("   " + "-" * 65)

for name, circ_in in test_states.items():
    # Ideal output
    circ_ideal = circ_in.copy()
    circ_ideal.unitary(ideal_gate, [0])
    ideal_out = Statevector.from_instruction(circ_ideal)
    
    # Actual output
    circ_actual = circ_in.copy()
    circ_actual.unitary(actual_gate, [0])
    actual_out = Statevector.from_instruction(circ_actual)
    
    # Compute overlap (fidelity)
    fidelity = abs(ideal_out.inner(actual_out))**2
    
    print(f"   {name:4s} | {str(ideal_out)[:20]:20s} | {str(actual_out)[:20]:20s} | F={fidelity:.4f}")

print("\n4. PROCESS TOMOGRAPHY WOULD:")
print("   - Measure all these outputs with many shots")
print("   - Reconstruct the process matrix χ_actual")
print("   - Compare with χ_ideal to identify the error")
print("   - Calculate process fidelity: F_proc = Tr(χ_ideal · χ_actual)")
print("   - Extract error parameters (rotation angle, etc.)")

print("\n" + "="*70)
print("This demonstrates how QPT can detect and quantify coherent errors")
print("="*70)
````

### Key Takeaways

1. **Quantum Process Tomography** reconstructs unknown quantum operations by measuring their effect on complete sets of input states.

2. **Purpose**: Characterize gates, identify errors, calibrate hardware, and benchmark devices.

3. **Process**: Prepare input states → Apply process → Measure outputs → Reconstruct process matrix.

4. **Applications**: 
   - Gate calibration and error correction
   - Device benchmarking
   - Error model construction
   - Coherent error identification

5. **Challenges**: Exponential scaling with qubit number, SPAM errors, statistical uncertainty.

6. **Connection to Error Correction**: QPT helps identify coherent errors that can be corrected through calibration, complementing QEC codes that handle incoherent errors.

---

### Summary: Process Tomography vs Error Correction

- **Process Tomography**: Characterizes what errors exist (diagnostic tool)
- **Quantum Error Correction**: Protects against errors (corrective tool)
- **Together**: QPT identifies errors → QEC corrects them


## 3 Qubit bit flip repetition code

### Overview

The **3-qubit bit flip repetition code** is the simplest quantum error correction code. It protects a single logical qubit against bit-flip errors by encoding it into three physical qubits.

### Key Concepts

#### 1. **Encoding**
- **Logical |0⟩** → **|000⟩** (three physical qubits in state |0⟩)
- **Logical |1⟩** → **|111⟩** (three physical qubits in state |1⟩)
- This creates redundancy: the same information is stored in three copies

#### 2. **Error Model**
- **Bit-flip error (X error)**: |0⟩ ↔ |1⟩
- The code can detect and correct **one bit-flip error** on any of the three qubits
- If two or more qubits flip, the code fails (cannot distinguish from a single flip)

#### 3. **Syndrome Measurement**
- **Syndrome**: A measurement that detects errors **without** measuring the logical qubit directly
- We use **parity checks** (CNOT gates + measurement) to compare qubits pairwise:
  - Compare qubit 0 and qubit 1 → parity bit 1
  - Compare qubit 1 and qubit 2 → parity bit 2
- The syndrome tells us **which qubit** has an error, but **not** the logical state

#### 4. **Syndrome Table**
| Syndrome (parity bits) | Error Location | Correction |
|------------------------|----------------|------------|
| 00 | No error | None |
| 01 | Qubit 2 flipped | Apply X to qubit 2 |
| 10 | Qubit 0 flipped | Apply X to qubit 0 |
| 11 | Qubit 1 flipped | Apply X to qubit 1 |

#### 5. **Error Correction Process**
1. **Encode**: Prepare |000⟩ or |111⟩
2. **Error occurs**: One qubit may flip (e.g., |000⟩ → |010⟩)
3. **Syndrome extraction**: Measure parity without collapsing the logical state
4. **Correction**: Apply X gate to the qubit indicated by the syndrome
5. **Decode**: Recover the original logical qubit

### Limitations

- ❌ **Cannot correct phase-flip errors** (Z errors)
- ❌ **Cannot correct two simultaneous bit-flip errors**
- ❌ **Cannot correct combined errors** (e.g., X and Z together)

#### 5. **Error Correction Process**
1. **Encode**: Prepare |000⟩ or |111⟩
2. **Error occurs**: One qubit may flip (e.g., |000⟩ → |010⟩)
3. **Syndrome extraction**: Measure parity without collapsing the logical state
4. **Correction**: Apply X gate to the qubit indicated by the syndrome
5. **Decode**: Recover the original logical qubit

### Limitations

- ❌ **Cannot correct phase-flip errors** (Z errors)
- ❌ **Cannot correct two simultaneous bit-flip errors**
- ❌ **Cannot correct combined errors** (e.g., X and Z together)

### Circuit Structure

```
Encoding:
|ψ⟩ ──[H]──●──●── |ψ₀⟩
|0⟩ ──────X──┼── |ψ₁⟩

Syndrome Measurement (using ancilla qubits):
|ψ₀⟩ ──●───────
|ψ₁⟩ ──X──●────

|0⟩  ──⊕─────── ancilla 1 (parity of qubits 0,1)
|0⟩  ──────⊕─── ancilla 2 (parity of qubits 1,2)

Correction:
Apply X gate based on syndrome measurement

Decoding:
|ψ₀⟩ ──●──●──
|ψ₁⟩ ──X──┼──
|ψ₂⟩ ─────X──
           |
         |ψ⟩ (recovered)
```

## 3 Qubit Phase Flip Repetition Code



### Overview

The **3-qubit phase flip repetition code** protects a single logical qubit against **phase-flip errors (Z errors)** by encoding it into three physical qubits. It's the dual of the bit flip code - while the bit flip code protects against X errors, this code protects against Z errors.

### Key Concepts

#### 1. **Encoding**
- **Logical |0⟩** → **|+++⟩** = (|0⟩+|1⟩)/√2 ⊗ (|0⟩+|1⟩)/√2 ⊗ (|0⟩+|1⟩)/√2
- **Logical |1⟩** → **|---⟩** = (|0⟩-|1⟩)/√2 ⊗ (|0⟩-|1⟩)/√2 ⊗ (|0⟩-|1⟩)/√2
- Uses Hadamard gates to transform into the X-basis (|+⟩, |-⟩)

#### 2. **Error Model**
- **Phase-flip error (Z error)**: |+⟩ ↔ |-⟩ (flips phase, not bit)
- The code can detect and correct **one phase-flip error** on any of the three qubits
- Z|+⟩ = |-⟩ and Z|-⟩ = |+⟩

#### 3. **Syndrome Measurement**
- Similar to bit flip code, but works in the X-basis
- Uses CNOT gates with Hadamard transforms to measure parity in X-basis
- The syndrome tells us **which qubit** has a phase error

#### 4. **Syndrome Table**
| Syndrome (parity bits) | Error Location | Correction |
|------------------------|----------------|------------|
| 00 | No error | None |
| 01 | Qubit 2 phase-flipped | Apply Z to qubit 2 |
| 10 | Qubit 0 phase-flipped | Apply Z to qubit 0 |
| 11 | Qubit 1 phase-flipped | Apply Z to qubit 1 |

#### 5. **Error Correction Process**
1. **Encode**: Prepare |+++⟩ or |---⟩ using H gates
2. **Error occurs**: One qubit may phase-flip (e.g., |+++⟩ → |+-+⟩)
3. **Syndrome extraction**: Measure parity in X-basis
4. **Correction**: Apply Z gate to the qubit indicated by the syndrome
5. **Decode**: Recover the original logical qubit

### Key Difference from Bit Flip Code

- **Bit Flip Code**: Works in Z-basis (|0⟩, |1⟩), corrects X errors
- **Phase Flip Code**: Works in X-basis (|+⟩, |-⟩), corrects Z errors
- **Encoding**: Bit flip uses CNOT, phase flip uses H + CNOT
- **Correction**: Bit flip uses X gates, phase flip uses Z gates

### Limitations

- ❌ **Cannot correct bit-flip errors** (X errors)
- ❌ **Cannot correct two simultaneous phase-flip errors**
- ❌ **Cannot correct combined errors** (e.g., X and Z together)


````python
# Import required libraries (if not already imported)
from qiskit import QuantumCircuit, QuantumRegister, ClassicalRegister, transpile
from qiskit.quantum_info import Statevector, Operator
from qiskit_aer import AerSimulator
import numpy as np
import matplotlib.pyplot as plt

print("Libraries ready for Phase Flip Repetition Code!")
````

### Step 1: Encoding Circuit for Phase Flip Code

The encoding uses Hadamard gates to transform into the X-basis, then CNOT gates to create redundancy.


````python
def create_phase_flip_encoding_circuit(logical_state='0'):
    """
    Creates the encoding circuit for 3-qubit phase flip repetition code.
    
    Encoding:
    - |0⟩ → |+++⟩ = H|0⟩ ⊗ H|0⟩ ⊗ H|0⟩ (with CNOTs)
    - |1⟩ → |---⟩ = H|1⟩ ⊗ H|1⟩ ⊗ H|1⟩ (with CNOTs)
    
    Args:
        logical_state: '0' for |0⟩ or '1' for |1⟩
    
    Returns:
        QuantumCircuit: The encoding circuit
    """
    # Create quantum register with 3 qubits
    qreg = QuantumRegister(3, 'q')
    circ = QuantumCircuit(qreg)
    
    # Prepare the logical qubit
    if logical_state == '1':
        circ.x(0)  # Prepare |1⟩
    
    # Encoding for phase flip code:
    # 1. Apply Hadamard to all qubits (transform to X-basis)
    circ.h(0)
    circ.h(1)
    circ.h(2)
    
    # 2. Apply CNOT gates to create redundancy
    circ.cx(0, 1)
    circ.cx(0, 2)
    
    return circ

# Create encoding circuit for |0⟩ → |+++⟩
phase_encoding_0 = create_phase_flip_encoding_circuit('0')
print("Encoding circuit for logical |0⟩ → |+++⟩:")
print(phase_encoding_0)

# Visualize the circuit
fig = phase_encoding_0.draw('mpl', style='iqp')
fig.suptitle("Phase Flip Encoding: |0⟩ → |+++⟩", fontsize=14)
plt.show()

# Verify the encoding
state_after_encoding = Statevector.from_instruction(phase_encoding_0)
print(f"\nState after encoding |0⟩: {state_after_encoding}")
print(f"Expected: Equal superposition of |000⟩, |001⟩, |010⟩, |011⟩, |100⟩, |101⟩, |110⟩, |111⟩")
print(f"This represents |+++⟩ in computational basis")

# Create encoding circuit for |1⟩ → |---⟩
phase_encoding_1 = create_phase_flip_encoding_circuit('1')
print("\n\nEncoding circuit for logical |1⟩ → |---⟩:")
print(phase_encoding_1)

# Verify the encoding
state_after_encoding_1 = Statevector.from_instruction(phase_encoding_1)
print(f"\nState after encoding |1⟩: {state_after_encoding_1}")
print(f"Expected: Equal superposition with alternating signs (represents |---⟩)")
````

### Step 2: Syndrome Measurement for Phase Flip Code

For phase flip errors, we need to measure parity in the X-basis. This requires Hadamard gates before and after the CNOT operations.


````python
def create_phase_flip_syndrome_circuit():
    """
    Creates the syndrome measurement circuit for phase flip code.
    Uses Hadamard gates to measure in X-basis.
    
    Returns:
        QuantumCircuit: The syndrome measurement circuit
    """
    # 3 data qubits + 2 ancilla qubits
    qreg_data = QuantumRegister(3, 'data')
    qreg_ancilla = QuantumRegister(2, 'ancilla')
    creg_syndrome = ClassicalRegister(2, 'syndrome')
    
    circ = QuantumCircuit(qreg_data, qreg_ancilla, creg_syndrome)
    
    # To measure parity in X-basis, we need to:
    # 1. Transform to Z-basis with H gates
    # 2. Measure parity with CNOT
    # 3. Transform back to X-basis
    
    # Measure parity of qubits 0 and 1 → ancilla 0
    circ.h(0)  # Transform qubit 0 to Z-basis
    circ.h(1)  # Transform qubit 1 to Z-basis
    circ.cx(0, 3)  # CNOT from data qubit 0 to ancilla 0
    circ.cx(1, 3)  # CNOT from data qubit 1 to ancilla 0
    circ.h(0)  # Transform back to X-basis
    circ.h(1)  # Transform back to X-basis
    
    # Measure parity of qubits 1 and 2 → ancilla 1
    circ.h(1)  # Transform qubit 1 to Z-basis
    circ.h(2)  # Transform qubit 2 to Z-basis
    circ.cx(1, 4)  # CNOT from data qubit 1 to ancilla 1
    circ.cx(2, 4)  # CNOT from data qubit 2 to ancilla 1
    circ.h(1)  # Transform back to X-basis
    circ.h(2)  # Transform back to X-basis
    
    # Measure ancilla qubits
    circ.measure(qreg_ancilla, creg_syndrome)
    
    return circ

phase_syndrome_circuit = create_phase_flip_syndrome_circuit()
print("Phase flip syndrome measurement circuit:")
print(phase_syndrome_circuit)

fig = phase_syndrome_circuit.draw('mpl', style='iqp')
fig.suptitle("Phase Flip Syndrome Measurement Circuit", fontsize=14)
plt.show()

# Explain the syndrome table
print("\n" + "="*60)
print("PHASE FLIP SYNDROME TABLE")
print("="*60)
print("Syndrome (ancilla[1], ancilla[0]) | Error Location | Correction")
print("-" * 60)
print("00                              | No error       | None")
print("01                              | Qubit 2        | Apply Z to qubit 2")
print("10                              | Qubit 0        | Apply Z to qubit 0")
print("11                              | Qubit 1        | Apply Z to qubit 1")
print("="*60)
````

### Step 3: Complete Phase Flip Error Correction Protocol

Let's put it all together and test the complete protocol.


````python
def complete_phase_flip_error_correction_protocol(logical_state='0', error_qubit=None, shots=1024):
    """
    Complete 3-qubit phase flip repetition code protocol.
    
    Args:
        logical_state: '0' or '1' - the logical qubit to encode
        error_qubit: None, 0, 1, or 2 - which qubit to apply Z error to
        shots: Number of measurement shots
    
    Returns:
        QuantumCircuit: The complete error correction circuit
    """
    # Create registers
    qreg_data = QuantumRegister(3, 'data')
    qreg_ancilla = QuantumRegister(2, 'ancilla')
    creg_syndrome = ClassicalRegister(2, 'syndrome')
    creg_output = ClassicalRegister(1, 'output')
    
    # Create complete circuit
    circ = QuantumCircuit(qreg_data, qreg_ancilla, creg_syndrome, creg_output)
    
    # Step 1: Encoding (transform to X-basis and create redundancy)
    if logical_state == '1':
        circ.x(0)
    circ.h(0)
    circ.h(1)
    circ.h(2)
    circ.cx(0, 1)
    circ.cx(0, 2)
    
    # Step 2: Introduce error (if specified)
    if error_qubit is not None:
        circ.z(error_qubit)  # Phase-flip error (Z error)
        print(f"✓ Error introduced: Phase-flip (Z) on qubit {error_qubit}")
    
    # Step 3: Syndrome measurement (in X-basis)
    circ.barrier()
    # Parity of qubits 0 and 1
    circ.h(0)
    circ.h(1)
    circ.cx(0, 3)
    circ.cx(1, 3)
    circ.h(0)
    circ.h(1)
    # Parity of qubits 1 and 2
    circ.h(1)
    circ.h(2)
    circ.cx(1, 4)
    circ.cx(2, 4)
    circ.h(1)
    circ.h(2)
    circ.measure(qreg_ancilla, creg_syndrome)
    
    # Step 4: Error correction (classical feedback)
    circ.barrier()
    with circ.if_test((creg_syndrome, 1)):  # Syndrome 01 → correct qubit 2
        circ.z(2)

    with circ.if_test((creg_syndrome, 2)):  # Syndrome 10 → correct qubit 0
        circ.z(0)

    with circ.if_test((creg_syndrome, 3)):  # Syndrome 11 → correct qubit 1
        circ.z(1)
    
    # Step 5: Decoding (reverse encoding)
    circ.barrier()
    circ.cx(0, 2)
    circ.cx(0, 1)
    circ.h(0)
    circ.h(1)
    circ.h(2)
    
    # Step 6: Measure logical qubit
    circ.measure(0, creg_output)
    
    return circ

# Test Case 1: No error
print("="*70)
print("TEST CASE 1: Logical |0⟩ with NO PHASE ERROR")
print("="*70)
circ_no_error = complete_phase_flip_error_correction_protocol('0', error_qubit=None)
print("\nComplete circuit:")
print(circ_no_error)

# Execute
backend = AerSimulator()
job = backend.run(transpile(circ_no_error, backend), shots=1024)
result = job.result()
counts = result.get_counts()

print(f"\nResults: {counts}")
print(f"✓ Logical |0⟩ recovered: {counts.get('0 00', 0)/1024*100:.1f}%")
````

### Summary: Phase Flip vs Bit Flip Codes

**Key Differences:**

| Aspect | Bit Flip Code | Phase Flip Code |
|--------|---------------|-----------------|
| **Basis** | Z-basis (|0⟩, |1⟩) | X-basis (|+⟩, |-⟩) |
| **Encoding** | |0⟩ → |000⟩, |1⟩ → |111⟩ | |0⟩ → |+++⟩, |1⟩ → |---⟩ |

| **Error Type** | X errors (bit-flip) | Z errors (phase-flip) |
| **Correction** | Apply X gates | Apply Z gates |
| **Syndrome** | Measure in Z-basis | Measure in X-basis |

**Both codes are complementary:**
- Bit flip code protects against X errors
- Phase flip code protects against Z errors
- Together they form the foundation for codes that protect against both (like Shor code)

````python
# Import required libraries (if not already imported)
from qiskit import QuantumCircuit, QuantumRegister, ClassicalRegister, transpile
from qiskit.quantum_info import Statevector, Operator
from qiskit_aer import AerSimulator
import numpy as np
import matplotlib.pyplot as plt

print("Libraries ready for Shor Code implementation!")
````

### Step 1: Encoding Circuit for Shor Code

The encoding process:
1. Apply Hadamard to transform to X-basis (phase flip encoding)
2. Apply CNOT gates to create redundancy within each block (bit flip encoding)


````python
def create_shor_encoding_circuit(logical_state='0'):
    """
    Creates the encoding circuit for 9-qubit Shor code.
    
    Encoding process:
    1. Phase flip encoding: |0⟩ → |+++⟩, |1⟩ → |---⟩
    2. Bit flip encoding: Each |+⟩/|-⟩ → |000⟩/|111⟩
    
    Args:
        logical_state: '0' for |0⟩ or '1' for |1⟩
    
    Returns:
        QuantumCircuit: The encoding circuit
    """
    # Create quantum register with 9 qubits
    qreg = QuantumRegister(9, 'q')
    circ = QuantumCircuit(qreg)
    
    # Prepare the logical qubit
    if logical_state == '1':
        circ.x(0)  # Prepare |1⟩
    
    # Step 1: Phase flip encoding (transform to X-basis)
    # Apply Hadamard to qubits 0, 3, 6 (first qubit of each block)
    circ.h(0)
    circ.h(3)
    circ.h(6)
    
    # Step 2: Bit flip encoding within each block
    # Block 1: qubits 0, 1, 2
    circ.cx(0, 1)
    circ.cx(0, 2)
    
    # Block 2: qubits 3, 4, 5
    circ.cx(3, 4)
    circ.cx(3, 5)
    
    # Block 3: qubits 6, 7, 8
    circ.cx(6, 7)
    circ.cx(6, 8)
    
    # Step 3: Phase flip encoding between blocks
    # CNOT from block 1 to block 2 and block 3
    circ.cx(0, 3)
    circ.cx(0, 6)
    
    return circ

# Create encoding circuit for |0⟩
shor_encoding_0 = create_shor_encoding_circuit('0')
print("Shor code encoding circuit for logical |0⟩:")
print(shor_encoding_0)

# Visualize the circuit
fig = shor_encoding_0.draw('mpl', style='iqp', scale=0.6)
fig.suptitle("Shor Code Encoding: |0⟩ → 9-qubit encoded state", fontsize=14)
plt.show()

# Verify the encoding
state_after_encoding = Statevector.from_instruction(shor_encoding_0)
print(f"\nState after encoding |0⟩:")
print(f"Number of basis states: {len(state_after_encoding)}")
print(f"Dimensions: {state_after_encoding.dims}")

# Show some non-zero amplitudes
print("\nSample of encoded state (showing first few non-zero amplitudes):")
non_zero_count = 0
for i, amp in enumerate(state_after_encoding):
    if abs(amp) > 1e-6 and non_zero_count < 5:
        binary = format(i, '09b')  # 9 qubits
        print(f"  |{binary}⟩: {amp:.4f} (probability: {abs(amp)**2:.4f})")
        non_zero_count += 1
print(f"\nTotal non-zero amplitudes: {sum(1 for amp in state_after_encoding if abs(amp) > 1e-6)}")
print("(The encoded state is a superposition of many basis states)")
````

### Step 2: Syndrome Measurement for Shor Code

The Shor code requires two types of syndrome measurements:
1. **Bit-flip syndrome**: Within each block of 3 qubits
2. **Phase-flip syndrome**: Between blocks


````python
def create_shor_syndrome_circuit():
    """
    Creates the syndrome measurement circuit for Shor code.
    
    Syndrome measurements:
    1. Bit-flip syndrome: 6 ancilla qubits (2 per block × 3 blocks)
    2. Phase-flip syndrome: 2 ancilla qubits (between blocks)
    
    Returns:
        QuantumCircuit: The syndrome measurement circuit
    """
    # 9 data qubits + 8 ancilla qubits (6 for bit-flip + 2 for phase-flip)
    qreg_data = QuantumRegister(9, 'data')
    qreg_ancilla_bf = QuantumRegister(6, 'ancilla_bf')  # Bit-flip ancillas
    qreg_ancilla_pf = QuantumRegister(2, 'ancilla_pf')  # Phase-flip ancillas
    creg_syndrome = ClassicalRegister(8, 'syndrome')  # 8 syndrome bits
    
    circ = QuantumCircuit(qreg_data, qreg_ancilla_bf, qreg_ancilla_pf, creg_syndrome)
    
    # BIT-FLIP SYNDROME MEASUREMENT (within each block)
    # Block 1: qubits 0, 1, 2
    circ.cx(0, 9)   # Parity of qubits 0 and 1 → ancilla_bf[0]
    circ.cx(1, 9)
    circ.cx(1, 10)  # Parity of qubits 1 and 2 → ancilla_bf[1]
    circ.cx(2, 10)
    
    # Block 2: qubits 3, 4, 5
    circ.cx(3, 11)  # Parity of qubits 3 and 4 → ancilla_bf[2]
    circ.cx(4, 11)
    circ.cx(4, 12)  # Parity of qubits 4 and 5 → ancilla_bf[3]
    circ.cx(5, 12)
    
    # Block 3: qubits 6, 7, 8
    circ.cx(6, 13)  # Parity of qubits 6 and 7 → ancilla_bf[4]
    circ.cx(7, 13)
    circ.cx(7, 14)  # Parity of qubits 7 and 8 → ancilla_bf[5]
    circ.cx(8, 14)
    
    # PHASE-FLIP SYNDROME MEASUREMENT (between blocks)
    # Need to measure in X-basis, so use Hadamard gates
    # Compare block 1 with block 2
    circ.h(0)
    circ.h(1)
    circ.h(2)  # Transform block 1 to Z-basis
    circ.h(3)
    circ.h(4)
    circ.h(5)  # Transform block 2 to Z-basis
    circ.cx(0, 15)  # Parity of blocks 1 and 2 → ancilla_pf[0]
    circ.cx(3, 15)
    circ.h(0)
    circ.h(1)
    circ.h(2)  # Transform back
    circ.h(3)
    circ.h(4)
    circ.h(5)
    
    # Compare block 2 with block 3
    circ.h(3)
    circ.h(4)
    circ.h(5)  # Transform block 2 to Z-basis
    circ.h(6)
    circ.h(7)
    circ.h(8)  # Transform block 3 to Z-basis
    circ.cx(3, 16)  # Parity of blocks 2 and 3 → ancilla_pf[1]
    circ.cx(6, 16)
    circ.h(3)
    circ.h(4)
    circ.h(5)  # Transform back
    circ.h(6)
    circ.h(7)
    circ.h(8)
    
    # Measure all ancilla qubits
    circ.measure(qreg_ancilla_bf, creg_syndrome[0:6])  # Bit-flip syndromes
    circ.measure(qreg_ancilla_pf, creg_syndrome[6:8])  # Phase-flip syndromes
    
    return circ

shor_syndrome_circuit = create_shor_syndrome_circuit()
print("Shor code syndrome measurement circuit:")
print(shor_syndrome_circuit)

print("\n" + "="*70)
print("SHOR CODE SYNDROME STRUCTURE")
print("="*70)
print("Syndrome bits [0-5]: Bit-flip syndromes (2 per block × 3 blocks)")
print("  - Bits [0,1]: Block 1 (qubits 0,1,2)")
print("  - Bits [2,3]: Block 2 (qubits 3,4,5)")
print("  - Bits [4,5]: Block 3 (qubits 6,7,8)")
print("\nSyndrome bits [6-7]: Phase-flip syndromes (between blocks)")
print("  - Bit 6: Parity between blocks 1 and 2")
print("  - Bit 7: Parity between blocks 2 and 3")
print("="*70)
````

````python
def complete_shor_code_protocol(logical_state='0', error_type=None, error_qubit=None, shots=1024):
    """
    Complete 9-qubit Shor code protocol.
    
    Args:
        logical_state: '0' or '1' - the logical qubit to encode
        error_type: None, 'X', 'Z', or 'Y' - type of error
        error_qubit: None or 0-8 - which qubit to apply error to
        shots: Number of measurement shots
    
    Returns:
        QuantumCircuit: The complete error correction circuit
    """
    # Create registers
    qreg_data = QuantumRegister(9, 'data')
    qreg_ancilla_bf = QuantumRegister(6, 'ancilla_bf')
    qreg_ancilla_pf = QuantumRegister(2, 'ancilla_pf')
    creg_syndrome = ClassicalRegister(8, 'syndrome')
    creg_output = ClassicalRegister(1, 'output')
    
    circ = QuantumCircuit(qreg_data, qreg_ancilla_bf, qreg_ancilla_pf, creg_syndrome, creg_output)
    
    # Step 1: Encoding
    if logical_state == '1':
        circ.x(0)
    # Phase flip encoding
    circ.h(0)
    circ.h(3)
    circ.h(6)
    # Bit flip encoding within blocks
    circ.cx(0, 1)
    circ.cx(0, 2)
    circ.cx(3, 4)
    circ.cx(3, 5)
    circ.cx(6, 7)
    circ.cx(6, 8)
    # Phase flip encoding between blocks
    circ.cx(0, 3)
    circ.cx(0, 6)
    
    # Step 2: Introduce error (if specified)
    if error_qubit is not None and error_type is not None:
        if error_type == 'X':
            circ.x(error_qubit)
            print(f"✓ Error introduced: Bit-flip (X) on qubit {error_qubit}")
        elif error_type == 'Z':
            circ.z(error_qubit)
            print(f"✓ Error introduced: Phase-flip (Z) on qubit {error_qubit}")
        elif error_type == 'Y':
            circ.y(error_qubit)
            print(f"✓ Error introduced: Combined (Y) on qubit {error_qubit}")
    
    # Step 3: Syndrome measurement
    circ.barrier()
    
    # Bit-flip syndrome (within blocks)
    # Block 1
    circ.cx(0, 9)
    circ.cx(1, 9)
    circ.cx(1, 10)
    circ.cx(2, 10)
    # Block 2
    circ.cx(3, 11)
    circ.cx(4, 11)
    circ.cx(4, 12)
    circ.cx(5, 12)
    # Block 3
    circ.cx(6, 13)
    circ.cx(7, 13)
    circ.cx(7, 14)
    circ.cx(8, 14)
    
    # Phase-flip syndrome (between blocks)
    # Blocks 1 and 2
    circ.h(0)
    circ.h(1)
    circ.h(2)
    circ.h(3)
    circ.h(4)
    circ.h(5)
    circ.cx(0, 15)
    circ.cx(3, 15)
    circ.h(0)
    circ.h(1)
    circ.h(2)
    circ.h(3)
    circ.h(4)
    circ.h(5)
    # Blocks 2 and 3
    circ.h(3)
    circ.h(4)
    circ.h(5)
    circ.h(6)
    circ.h(7)
    circ.h(8)
    circ.cx(3, 16)
    circ.cx(6, 16)
    circ.h(3)
    circ.h(4)
    circ.h(5)
    circ.h(6)
    circ.h(7)
    circ.h(8)
    
    circ.measure(qreg_ancilla_bf, creg_syndrome[0:6])
    circ.measure(qreg_ancilla_pf, creg_syndrome[6:8])
    
    # Step 4: Error correction
    # Note: In a full implementation, we would decode the syndrome and apply corrections
    # For simplicity, we'll show the structure
    circ.barrier()
    
    # Step 5: Decoding (reverse encoding)
    circ.barrier()
    # Reverse phase flip encoding between blocks
    circ.cx(0, 6)
    circ.cx(0, 3)
    # Reverse bit flip encoding within blocks
    circ.cx(6, 8)
    circ.cx(6, 7)
    circ.cx(3, 5)
    circ.cx(3, 4)
    circ.cx(0, 2)
    circ.cx(0, 1)
    # Reverse phase flip encoding
    circ.h(6)
    circ.h(3)
    circ.h(0)
    
    # Step 6: Measure logical qubit
    circ.measure(0, creg_output)
    
    return circ

# Test Case 1: No error
print("="*70)
print("TEST CASE 1: Logical |0⟩ with NO ERROR")
print("="*70)
circ_no_error = complete_shor_code_protocol('0', error_type=None, error_qubit=None)
print("\nComplete circuit (first few gates shown):")
print(circ_no_error)

# Execute
backend = AerSimulator()
job = backend.run(transpile(circ_no_error, backend), shots=1024)
result = job.result()
counts = result.get_counts()

print(f"\nResults: {counts}")
print(f"✓ Logical |0⟩ recovered: {counts.get('0 ' + '0'*8, 0)/1024*100:.1f}%")
````

### Summary: Shor Code

**Key Features:**

1. **9-Qubit Code**: Encodes 1 logical qubit into 9 physical qubits
2. **Dual Protection**: Corrects both bit-flip (X) and phase-flip (Z) errors
3. **Nested Structure**: Combines bit flip and phase flip codes
4. **Error Threshold**: Can correct 1 error of any type (X, Z, or Y)

**Encoding Structure:**
- **Level 1**: Phase flip code → 3 qubits in X-basis
- **Level 2**: Bit flip code → Each qubit → 3 qubits
- **Total**: 3 × 3 = 9 qubits

**Syndrome Measurements:**
- **6 bits** for bit-flip errors (2 per block)
- **2 bits** for phase-flip errors (between blocks)
- **Total**: 8 syndrome bits

**Comparison with Repetition Codes:**

| Code | Qubits | Protects Against | Errors Corrected |
|------|--------|-------------------|------------------|
| Bit Flip (3-qubit) | 3 | X errors only | 1 X error |
| Phase Flip (3-qubit) | 3 | Z errors only | 1 Z error |
| **Shor Code (9-qubit)** | **9** | **X and Z errors** | **1 error (X, Z, or Y)** |

**The Shor code demonstrates how to combine simple codes to protect against multiple error types!**


### Overview

The **Shor code** (also known as the 9-qubit Shor code) is a quantum error correction code that protects a single logical qubit against **both bit-flip (X) and phase-flip (Z) errors**. It combines the 3-qubit bit flip code and the 3-qubit phase flip code in a nested structure.

**Key Feature**: Encodes **1 logical qubit** into **9 physical qubits** and can correct **1 error of any type** (X, Z, or Y).

### Structure

The Shor code uses a **two-level encoding**:

1. **First level**: Phase flip code (protects against Z errors)
   - Encodes logical qubit into 3 qubits in X-basis
   
2. **Second level**: Bit flip code (protects against X errors)
   - Each of the 3 qubits from level 1 is encoded into 3 qubits
   - Total: 3 × 3 = 9 qubits

### Encoding

**Logical |0⟩** → |+++⟩ → |000 000 000⟩ (after bit flip encoding)
**Logical |1⟩** → |---⟩ → |111 111 111⟩ (after bit flip encoding)

Where:
- |+++⟩ = (|0⟩+|1⟩)/√2 ⊗ (|0⟩+|1⟩)/√2 ⊗ (|0⟩+|1⟩)/√2
- |---⟩ = (|0⟩-|1⟩)/√2 ⊗ (|0⟩-|1⟩)/√2 ⊗ (|0⟩-|1⟩)/√2

### Error Correction Capabilities

The Shor code can correct:
- ✅ **1 bit-flip error** (X error) on any qubit
- ✅ **1 phase-flip error** (Z error) on any qubit  
- ✅ **1 combined error** (Y error = iXZ) on any qubit
- ❌ **2 or more errors** (cannot correct)

### Syndrome Measurement

The Shor code uses **two types of syndrome measurements**:

1. **Bit-flip syndrome**: Measures parity within each block of 3 qubits
   - 6 ancilla qubits (2 per block × 3 blocks)
   
2. **Phase-flip syndrome**: Measures parity between blocks
   - 2 ancilla qubits

### Advantages

- **First code** to correct both X and Z errors
- **Foundation** for more advanced codes
- **Demonstrates** nested encoding structure

### Limitations

- Requires **9 physical qubits** per logical qubit
- Cannot correct **2 or more errors**
- More complex than repetition codes

````python
def create_steane_encoding_circuit(logical_state='0'):
    """
    Creates the encoding circuit for 7-qubit Steane code.
    
    The encoding creates a superposition of classical Hamming codewords:
    - |0_L⟩: Superposition of even-weight codewords
    - |1_L⟩: Superposition of odd-weight codewords
    
    Args:
        logical_state: '0' for |0_L⟩ or '1' for |1_L⟩
    
    Returns:
        QuantumCircuit: The encoding circuit
    """
    # Create quantum register with 7 qubits
    qreg = QuantumRegister(7, 'q')
    circ = QuantumCircuit(qreg)
    
    # Prepare the logical qubit
    if logical_state == '1':
        circ.x(0)  # Prepare |1⟩
    
    # Steane code encoding circuit
    # This creates the logical |0_L⟩ or |1_L⟩ state
    
    # Step 1: Create superposition
    circ.h(1)
    circ.h(2)
    circ.h(3)
    
    # Step 2: Apply CNOT gates based on Hamming code structure
    # These create the codeword structure
    circ.cx(1, 4)
    circ.cx(2, 4)
    circ.cx(1, 5)
    circ.cx(3, 5)
    circ.cx(2, 6)
    circ.cx(3, 6)
    
    # Step 3: Encode logical information
    circ.cx(0, 4)
    circ.cx(0, 5)
    circ.cx(0, 6)
    
    return circ

# Create encoding circuit for |0_L⟩
steane_encoding_0 = create_steane_encoding_circuit('0')
print("Steane code encoding circuit for logical |0_L⟩:")
print(steane_encoding_0)

# Visualize the circuit
fig = steane_encoding_0.draw('mpl', style='iqp', scale=0.7)
fig.suptitle("Steane Code Encoding: |0⟩ → |0_L⟩ (7-qubit encoded state)", fontsize=14)
plt.show()

# Verify the encoding
state_after_encoding = Statevector.from_instruction(steane_encoding_0)
print(f"\nState after encoding |0⟩:")
print(f"Number of basis states: {len(state_after_encoding)}")
print(f"Dimensions: {state_after_encoding.dims}")

# Show some non-zero amplitudes
print("\nSample of encoded state (showing first few non-zero amplitudes):")
non_zero_count = 0
for i, amp in enumerate(state_after_encoding):
    if abs(amp) > 1e-6 and non_zero_count < 8:
        binary = format(i, '07b')  # 7 qubits
        weight = binary.count('1')  # Hamming weight
        print(f"  |{binary}⟩ (weight {weight}): {amp:.4f} (probability: {abs(amp)**2:.4f})")
        non_zero_count += 1
print(f"\nTotal non-zero amplitudes: {sum(1 for amp in state_after_encoding if abs(amp) > 1e-6)}")
print("(The encoded state is a superposition of Hamming codewords)")
````

### Step 3: Complete Steane Code Error Correction Protocol

Let's implement the complete protocol with error correction.


````python
def complete_steane_code_protocol(logical_state='0', error_type=None, error_qubit=None, shots=1024):
    """
    Complete 7-qubit Steane code protocol.
    
    Args:
        logical_state: '0' or '1' - the logical qubit to encode
        error_type: None, 'X', 'Z', or 'Y' - type of error
        error_qubit: None or 0-6 - which qubit to apply error to
        shots: Number of measurement shots
    
    Returns:
        QuantumCircuit: The complete error correction circuit
    """
    # Create registers
    qreg_data = QuantumRegister(7, 'data')
    qreg_ancilla = QuantumRegister(6, 'ancilla')
    creg_syndrome = ClassicalRegister(6, 'syndrome')
    creg_output = ClassicalRegister(1, 'output')
    
    circ = QuantumCircuit(qreg_data, qreg_ancilla, creg_syndrome, creg_output)
    
    # Step 1: Encoding
    if logical_state == '1':
        circ.x(0)
    # Create superposition
    circ.h(1)
    circ.h(2)
    circ.h(3)
    # Apply CNOT gates for Hamming code structure
    circ.cx(1, 4)
    circ.cx(2, 4)
    circ.cx(1, 5)
    circ.cx(3, 5)
    circ.cx(2, 6)
    circ.cx(3, 6)
    # Encode logical information
    circ.cx(0, 4)
    circ.cx(0, 5)
    circ.cx(0, 6)
    
    # Step 2: Introduce error (if specified)
    if error_qubit is not None and error_type is not None:
        if error_type == 'X':
            circ.x(error_qubit)
            print(f"✓ Error introduced: Bit-flip (X) on qubit {error_qubit}")
        elif error_type == 'Z':
            circ.z(error_qubit)
            print(f"✓ Error introduced: Phase-flip (Z) on qubit {error_qubit}")
        elif error_type == 'Y':
            circ.y(error_qubit)
            print(f"✓ Error introduced: Combined (Y) on qubit {error_qubit}")
    
    # Step 3: Syndrome measurement
    circ.barrier()
    
    # Measure X-stabilizers (detect Z errors)
    # g₁ = X₀ X₁ X₂ X₄
    circ.h(7)
    circ.cx(0, 7)
    circ.cx(1, 7)
    circ.cx(2, 7)
    circ.cx(4, 7)
    circ.h(7)
    circ.measure(7, 0)
    
    # g₂ = X₁ X₂ X₃ X₅
    circ.h(8)
    circ.cx(1, 8)
    circ.cx(2, 8)
    circ.cx(3, 8)
    circ.cx(5, 8)
    circ.h(8)
    circ.measure(8, 1)
    
    # g₃ = X₂ X₃ X₄ X₆
    circ.h(9)
    circ.cx(2, 9)
    circ.cx(3, 9)
    circ.cx(4, 9)
    circ.cx(6, 9)
    circ.h(9)
    circ.measure(9, 2)
    
    # Measure Z-stabilizers (detect X errors)
    # g₄ = Z₀ Z₁ Z₂ Z₄
    circ.cx(0, 10)
    circ.cx(1, 10)
    circ.cx(2, 10)
    circ.cx(4, 10)
    circ.measure(10, 3)
    
    # g₅ = Z₁ Z₂ Z₃ Z₅
    circ.cx(1, 11)
    circ.cx(2, 11)
    circ.cx(3, 11)
    circ.cx(5, 11)
    circ.measure(11, 4)
    
    # g₆ = Z₂ Z₃ Z₄ Z₆
    circ.cx(2, 12)
    circ.cx(3, 12)
    circ.cx(4, 12)
    circ.cx(6, 12)
    circ.measure(12, 5)
    
    # Step 4: Error correction
    # Note: In a full implementation, we would decode the syndrome and apply corrections
    # For simplicity, we'll show the structure
    circ.barrier()
    
    # Step 5: Decoding (reverse encoding)
    circ.barrier()
    # Reverse encoding
    circ.cx(0, 6)
    circ.cx(0, 5)
    circ.cx(0, 4)
    circ.cx(3, 6)
    circ.cx(2, 6)
    circ.cx(3, 5)
    circ.cx(1, 5)
    circ.cx(2, 4)
    circ.cx(1, 4)
    circ.h(3)
    circ.h(2)
    circ.h(1)
    
    # Step 6: Measure logical qubit
    circ.measure(0, creg_output)
    
    return circ

# Test Case 1: No error
print("="*70)
print("TEST CASE 1: Logical |0_L⟩ with NO ERROR")
print("="*70)
circ_no_error = complete_steane_code_protocol('0', error_type=None, error_qubit=None)
print("\nComplete circuit created")

# Execute
backend = AerSimulator()
job = backend.run(transpile(circ_no_error, backend), shots=1024)
result = job.result()
counts = result.get_counts()

print(f"\nResults (showing most common outcomes):")
sorted_counts = sorted(counts.items(), key=lambda x: x[1], reverse=True)[:3]
for key, value in sorted_counts:
    parts = key.split()
    if len(parts) >= 2:
        output = parts[0]
        syndrome = parts[1]
        print(f"  Output {output}, Syndrome {syndrome}: {value/1024*100:.1f}%")

print(f"\n✓ Expected: Syndrome 000000 (no errors)")
print(f"✓ Most common outcome: {sorted_counts[0][0]}")
````

### Overview

The **Steane code** (also known as the 7-qubit CSS code) is a quantum error correction code that protects a single logical qubit against **both bit-flip (X) and phase-flip (Z) errors**. It's a **CSS (Calderbank-Shor-Steane) code** built from classical Hamming codes.

**Key Feature**: Encodes **1 logical qubit** into **7 physical qubits** and can correct **1 error of any type** (X, Z, or Y).

### Structure

The Steane code is based on the **classical [7,4,3] Hamming code**:
- **7 bits** total
- **4 information bits**
- **3 parity bits**
- **Distance 3** (can detect 2 errors, correct 1)

### CSS Code Structure

CSS codes use **two classical codes**:
1. **C₁**: Handles bit-flip errors (X errors)
2. **C₂**: Handles phase-flip errors (Z errors)
3. **C₁ and C₂ are dual**: C₂ = C₁⊥ (dual code)

For Steane code:
- Both C₁ and C₂ are the [7,4,3] Hamming code
- C₁ = C₂ = C₂⊥ (self-dual property)

### Encoding

The Steane code encodes:
- **Logical |0_L⟩**: Superposition of all even-weight codewords
- **Logical |1_L⟩**: Superposition of all odd-weight codewords

The codewords are:
- **Even-weight**: 0000000, 1101001, 1011010, 0110011, 0111100, 1010101, 1100110, 0001111
- **Odd-weight**: 1111111, 0010110, 0100101, 1001100, 1000011, 0101010, 0011001, 1110000

### Stabilizers

The Steane code has **6 stabilizer generators**:

**X-stabilizers** (detect Z errors):
- g₁ = X₀ X₁ X₂ X₄
- g₂ = X₁ X₂ X₃ X₅
- g₃ = X₂ X₃ X₄ X₆

**Z-stabilizers** (detect X errors):
- g₄ = Z₀ Z₁ Z₂ Z₄
- g₅ = Z₁ Z₂ Z₃ Z₅
- g₆ = Z₂ Z₃ Z₄ Z₆

### Error Correction Capabilities

The Steane code can correct:
- ✅ **1 bit-flip error** (X error) on any qubit
- ✅ **1 phase-flip error** (Z error) on any qubit  
- ✅ **1 combined error** (Y error = iXZ) on any qubit
- ❌ **2 or more errors** (cannot correct)

### Syndrome Measurement

The Steane code uses **6 syndrome bits**:
- **3 bits** for X-stabilizers (detect Z errors)
- **3 bits** for Z-stabilizers (detect X errors)

### Advantages

- **More efficient** than Shor code (7 qubits vs 9 qubits)
- **Based on classical codes** (easier to understand)
- **CSS structure** (simplifies implementation)
- **Foundation** for fault-tolerant quantum computing

### Limitations

- Requires **7 physical qubits** per logical qubit
- Cannot correct **2 or more errors**
- More complex than repetition codes


## [n, k, d] Representation

### Overview

The **[n, k, d] representation** is a standard notation used to describe quantum error correction codes. It provides a compact way to specify the key parameters of a code.

### Parameters

- **n**: Total number of physical qubits
- **k**: Number of logical qubits (information qubits)
- **d**: Code distance (minimum number of qubits that must be flipped to convert one codeword into another)

### Understanding Each Parameter

#### **n (Total Physical Qubits)**
- The total number of physical qubits used in the code
- Includes both data qubits and ancilla qubits (for syndrome measurement)
- Larger n generally means better error correction, but requires more resources

#### **k (Logical Qubits)**
- The number of logical (protected) qubits encoded
- Represents the actual quantum information being protected
- For most simple codes: k = 1 (one logical qubit)

#### **d (Code Distance)**
- The **minimum Hamming distance** between any two distinct codewords
- Determines the error correction capability
- **Error detection**: Can detect up to (d-1) errors
- **Error correction**: Can correct up to ⌊(d-1)/2⌋ errors

### Distance Formula

The maximum number of errors a code can correct is:

**t = ⌊(d-1)/2⌋**

Where:
- **t** = number of correctable errors
- **d** = code distance
- ⌊⌋ = floor function (round down)

### Examples

#### **3-Qubit Bit Flip Code: [3, 1, 3]**
- **n = 3**: Uses 3 physical qubits
- **k = 1**: Encodes 1 logical qubit
- **d = 3**: Distance 3 (can correct 1 error)
- **t = ⌊(3-1)/2⌋ = ⌊2/2⌋ = 1** error

#### **7-Qubit Steane Code: [7, 1, 3]**
- **n = 7**: Uses 7 physical qubits
- **k = 1**: Encodes 1 logical qubit
- **d = 3**: Distance 3 (can correct 1 error)
- **t = ⌊(3-1)/2⌋ = 1** error

#### **9-Qubit Shor Code: [9, 1, 3]**
- **n = 9**: Uses 9 physical qubits
- **k = 1**: Encodes 1 logical qubit
- **d = 3**: Distance 3 (can correct 1 error)
- **t = ⌊(3-1)/2⌋ = 1** error

### Code Rate

The **code rate** is defined as:

**R = k/n**

- **R = 1/3** for [3, 1, 3] bit flip code
- **R = 1/7** for [7, 1, 3] Steane code
- **R = 1/9** for [9, 1, 3] Shor code

Lower code rate means more redundancy (better error correction) but requires more physical qubits.

### Distance and Error Correction

| Distance (d) | Detect Errors | Correct Errors | Formula |
|--------------|----------------|----------------|----------|
| 1 | 0 | 0 | No protection |
| 2 | 1 | 0 | Detection only |
| 3 | 2 | 1 | t = ⌊(3-1)/2⌋ = 1 |
| 4 | 3 | 1 | t = ⌊(4-1)/2⌋ = 1 |
| 5 | 4 | 2 | t = ⌊(5-1)/2⌋ = 2 |
| 7 | 6 | 3 | t = ⌊(7-1)/2⌋ = 3 |

### Quantum vs Classical Codes

**Classical codes**: [n, k, d] where k information bits are encoded into n bits with distance d.

**Quantum codes**: Similar notation, but:
- Protects against both bit-flip (X) and phase-flip (Z) errors
- Uses quantum superposition and entanglement
- Requires more sophisticated error correction techniques

### Key Takeaways

1. **Higher distance (d)** → **Better error correction** (can correct more errors)
2. **Higher n** → **More physical resources** required
3. **Higher k** → **More logical information** encoded
4. **Trade-off**: Better error correction requires more physical qubits (lower code rate)

### Summary

The **[n, k, d] representation** is essential for:
- **Comparing codes**: Which code is more efficient?
- **Understanding capabilities**: How many errors can it correct?
- **Resource planning**: How many qubits are needed?
- **Code design**: What parameters should we aim for?

## Stabilizers

### Overview

**Stabilizers** are a fundamental concept in quantum error correction. They are operators that "stabilize" (preserve) the code space - meaning they leave valid codewords unchanged. Stabilizers form the mathematical foundation for detecting and correcting errors in quantum codes.

### What are Stabilizers?

A **stabilizer** is a Pauli operator (or product of Pauli operators) that:
1. **Commutes** with all valid codewords
2. **Leaves codewords unchanged** when applied
3. **Detects errors** when they occur (errors anti-commute with stabilizers)

### Key Properties

1. **Stabilizer Group**: All stabilizers form a group under multiplication
2. **Code Space**: The set of states that are eigenstates of all stabilizers with eigenvalue +1
3. **Error Detection**: Errors that anti-commute with stabilizers are detected
4. **Syndrome**: The measurement outcomes of stabilizers form the error syndrome

### Mathematical Definition

For a quantum error correction code, the **stabilizer group S** is a set of Pauli operators such that:

- For any stabilizer **g ∈ S** and any codeword **|ψ⟩**:
  - **g|ψ⟩ = |ψ⟩** (stabilizer preserves the state)

- For any error **E** that anti-commutes with **g**:
  - **gE|ψ⟩ = -E|ψ⟩** (error is detected)

### Types of Stabilizers

#### **X-Stabilizers**
- Composed of **X (bit-flip) operators**
- Detect **Z (phase-flip) errors**
- Example: **X₀ X₁ X₂** (parity check in X-basis)

#### **Z-Stabilizers**
- Composed of **Z (phase-flip) operators**
- Detect **X (bit-flip) errors**
- Example: **Z₀ Z₁ Z₂** (parity check in Z-basis)

#### **Y-Stabilizers**
- Composed of **Y operators** (Y = iXZ)
- Can detect both X and Z errors
- Less commonly used in practice

### Stabilizer Generators

A **stabilizer generator** is a minimal set of stabilizers that generate the entire stabilizer group. For a code encoding **k** logical qubits into **n** physical qubits:

- Number of stabilizer generators = **n - k**
- These generators uniquely define the code space

### Examples

#### **3-Qubit Bit Flip Code**

**Stabilizers:**
- **g₁ = Z₀ Z₁** (parity of qubits 0 and 1)
- **g₂ = Z₁ Z₂** (parity of qubits 1 and 2)

**Code space:**
- **|000⟩** and **|111⟩** are eigenstates with eigenvalue +1
- Any single bit-flip error anti-commutes with at least one stabilizer

#### **7-Qubit Steane Code**

**X-Stabilizers (detect Z errors):**
- **g₁ = X₀ X₁ X₂ X₄**
- **g₂ = X₁ X₂ X₃ X₅**
- **g₃ = X₂ X₃ X₄ X₆**

**Z-Stabilizers (detect X errors):**
- **g₄ = Z₀ Z₁ Z₂ Z₄**
- **g₅ = Z₁ Z₂ Z₃ Z₅**
- **g₆ = Z₂ Z₃ Z₄ Z₆**

### Syndrome Measurement

**Syndrome measurement** is the process of measuring stabilizers to detect errors:

1. **Prepare ancilla qubit** in |+⟩ or |0⟩ state
2. **Apply controlled operations** based on stabilizer
3. **Measure ancilla** to get syndrome bit
4. **Syndrome pattern** indicates which error occurred

### Error Detection Mechanism

When an error **E** occurs:

1. **If E commutes with stabilizer g**: 
   - **gE|ψ⟩ = Eg|ψ⟩ = E|ψ⟩**
   - Error is **not detected** by this stabilizer

2. **If E anti-commutes with stabilizer g**:
   - **gE|ψ⟩ = -Eg|ψ⟩ = -E|ψ⟩**
   - Error is **detected** (syndrome bit = 1)

### Stabilizer Formalism

The **stabilizer formalism** is a powerful framework for:
- **Describing quantum codes** compactly
- **Analyzing error correction** capabilities
- **Designing new codes** systematically
- **Implementing fault-tolerant operations**

### Advantages

1. **Compact Description**: n-k generators describe 2^(n-k) stabilizers
2. **Systematic Approach**: Well-defined mathematical structure
3. **Error Detection**: Clear mechanism for detecting errors
4. **Code Design**: Systematic way to construct codes

### Limitations

1. **Requires Commutativity**: All stabilizers must commute
2. **Pauli Errors**: Best suited for Pauli error models
3. **Measurement Overhead**: Need ancilla qubits for syndrome measurement

### Key Takeaways

1. **Stabilizers define the code space** - states that are preserved
2. **Errors anti-commute with stabilizers** - this is how errors are detected
3. **Syndrome = measurement outcomes** - tells us which errors occurred
4. **Stabilizer generators** - minimal set needed to describe the code
5. **X-stabilizers detect Z errors, Z-stabilizers detect X errors**

### Summary

**Stabilizers are the mathematical foundation of quantum error correction codes. They provide:**
- **Code definition**: What states are valid codewords?
- **Error detection**: How do we detect errors?
- **Syndrome extraction**: How do we measure errors?
- **Systematic framework**: How do we design and analyze codes?

### Stabilizer Measurement Circuit

The process of measuring stabilizers involves:

1. **Prepare ancilla qubit** in appropriate state
2. **Apply controlled operations** (CNOT gates for Z-stabilizers, Hadamard + CNOT for X-stabilizers)
3. **Measure ancilla** to get syndrome bit
4. **Repeat** for all stabilizer generators

### Example: Measuring Z-Stabilizer

To measure **Z₀ Z₁** (parity check):

```
1. Prepare ancilla in |0⟩
2. Apply CNOT(0 → ancilla)
3. Apply CNOT(1 → ancilla)
4. Measure ancilla → gives parity (0 = even, 1 = odd)
```

### Example: Measuring X-Stabilizer

To measure **X₀ X₁** (parity check in X-basis):

```
1. Prepare ancilla in |+⟩ (apply H to |0⟩)
2. Apply CNOT(0 → ancilla)
3. Apply CNOT(1 → ancilla)
4. Apply H to ancilla
5. Measure ancilla → gives parity in X-basis
```

### Stabilizer Group Properties

1. **Closure**: Product of stabilizers is also a stabilizer
2. **Identity**: Identity operator I is always a stabilizer
3. **Inverse**: Every stabilizer has an inverse (itself, for Pauli operators)
4. **Commutativity**: All stabilizers commute with each other

### Code Space Dimension

For a code with **n** physical qubits and **k** logical qubits:
- **Number of stabilizer generators** = n - k
- **Code space dimension** = 2^k
- **Number of stabilizers** = 2^(n-k)

### Error Correction with Stabilizers

The error correction process:

1. **Encode**: Prepare state in code space (stabilized by all stabilizers)
2. **Error occurs**: State may leave code space
3. **Measure stabilizers**: Get syndrome (which stabilizers have eigenvalue -1)
4. **Decode syndrome**: Determine which error occurred
5. **Apply correction**: Restore state to code space

### Stabilizer Codes

**Stabilizer codes** are quantum error correction codes defined by their stabilizer group. Most important quantum codes are stabilizer codes:

- **Repetition codes** (bit flip, phase flip)
- **CSS codes** (Steane code, Shor code)
- **Surface codes** (topological codes)
- **Color codes**
- **LDPC codes**

### Advantages of Stabilizer Formalism

1. **Compact**: n-k generators describe entire code
2. **Systematic**: Well-defined mathematical structure
3. **Efficient**: Polynomial-time encoding/decoding
4. **General**: Applies to many important codes
5. **Fault-tolerant**: Can implement gates fault-tolerantly

### Conclusion

**Stabilizers are essential for:**
- **Defining quantum codes** mathematically
- **Detecting errors** through syndrome measurement
- **Understanding code structure** and properties
- **Designing new codes** systematically
- **Implementing fault-tolerant quantum computing**

**The stabilizer formalism provides the mathematical foundation for modern quantum error correction!**


## MLIR - Multi-Level Intermediate Representation 

### Overview

**MLIR (Multi-Level Intermediate Representation)** is a compiler infrastructure framework developed by **LLVM** (Low Level Virtual Machine). It provides a flexible, extensible system for building domain-specific compilers and optimizing code across different abstraction levels.

### What is MLIR?

MLIR is a **compiler infrastructure** that:
- Provides a **unified framework** for representing code at multiple abstraction levels
- Enables **progressive lowering** from high-level domain-specific languages to low-level machine code
- Supports **custom dialects** for different domains (AI/ML, quantum computing, hardware design, etc.)
- Facilitates **optimization** and **code generation** across different hardware targets

### Key Concepts

#### **1. Multi-Level Representation**
- MLIR allows representing code at **multiple abstraction levels** simultaneously
- High-level operations can coexist with low-level operations
- Enables **gradual transformation** from abstract to concrete representations

#### **2. Dialects**
- **Dialects** are namespaces for operations and types
- Each dialect defines its own set of operations, types, and attributes
- Examples: TensorFlow, PyTorch, Quantum, Affine, LLVM, GPU dialects

#### **3. Operations**
- **Operations** are the basic units of computation in MLIR
- Each operation has:
  - **Name**: Identifies the operation
  - **Operands**: Input values
  - **Results**: Output values
  - **Attributes**: Static metadata
  - **Regions**: Nested code blocks

#### **4. Types**
- MLIR has a **flexible type system**
- Types can be:
  - **Built-in types**: Integer, Float, Vector, Tensor
  - **Dialect-specific types**: Custom types defined by dialects
  - **Complex types**: Functions, MemRef, etc.

### Architecture

MLIR's architecture consists of:

1. **Core Infrastructure**
   - Operation representation
   - Type system
   - Attribute system
   - Region and block structure

2. **Dialect System**
   - Dialect registration
   - Operation definitions
   - Type definitions
   - Custom attributes

3. **Pass Infrastructure**
   - Transformation passes
   - Analysis passes
   - Pass pipelines
   - Pass managers

4. **Code Generation**
   - Target-specific backends
   - Lowering passes
   - Code emission

### Why MLIR?

#### **Problems with Traditional Compilers**

1. **Single IR Level**: Traditional compilers use one IR (Intermediate Representation)
2. **Rigid Structure**: Hard to extend for new domains
3. **Loss of Information**: High-level semantics lost during lowering
4. **Domain-Specific**: Each domain needs its own compiler

#### **MLIR Solutions**

1. **Multiple IR Levels**: Can represent code at many abstraction levels
2. **Extensible**: Easy to add new dialects and operations
3. **Preserves Semantics**: High-level information retained longer
4. **Unified Framework**: One infrastructure for many domains

### MLIR Dialects

#### **Built-in Dialects**

1. **Standard Dialect**
   - Basic operations (add, sub, mul, div)
   - Control flow (if, for, while)
   - Memory operations

2. **Affine Dialect**
   - Affine transformations
   - Loop optimizations
   - Polyhedral analysis

3. **LLVM Dialect**
   - LLVM IR operations
   - Low-level operations
   - Target code generation

4. **Tensor Dialect**
   - Tensor operations
   - Shape inference
   - Tensor transformations

#### **Domain-Specific Dialects**

1. **TensorFlow Dialect**
   - TensorFlow operations
   - Graph representation
   - TensorFlow-specific optimizations

2. **PyTorch Dialect**
   - PyTorch operations
   - Dynamic shapes
   - JIT compilation

3. **Quantum Dialect**
   - Quantum gates
   - Quantum circuits
   - Quantum-specific optimizations

4. **GPU Dialect**
   - GPU operations
   - Parallel execution
   - Memory management

### MLIR Workflow

The typical MLIR compilation workflow:

1. **Frontend**: Source code → MLIR (high-level dialect)
2. **Progressive Lowering**: High-level dialect → Lower-level dialects
3. **Optimization**: Apply optimization passes at each level
4. **Code Generation**: Lower to target-specific code (LLVM IR, assembly, etc.)

### Example: MLIR Code Structure

```mlir
// Example MLIR code
```

**Components:**
- `func`: Function operation
- `@add`: Function name
- `%arg0, %arg1`: Function arguments
- `i32`: 32-bit integer type

### MLIR Transformation Example

**Before (High-level):**
```mlir
func @compute(%x: tensor<10xf32>) -> tensor<10xf32> {
  %0 = "tensor.add"(%x, %x) : (tensor<10xf32>, tensor<10xf32>) -> tensor<10xf32>
  return %0 : tensor<10xf32>
}
```

**After (Lowered to Standard dialect):**
```mlir
func @compute(%x: memref<10xf32>) -> memref<10xf32> {
  %0 = memref.alloc() : memref<10xf32>
  affine.for %i = 0 to 10 {
    %1 = affine.load %x[%i] : memref<10xf32>
    %2 = arith.addf %1, %1 : f32
    affine.store %2, %0[%i] : memref<10xf32>
  }
  return %0 : memref<10xf32>
}
```

**Transformation:**
- High-level tensor operations → Low-level memory operations
- Implicit loops → Explicit loops
- Tensor types → Memory reference types
